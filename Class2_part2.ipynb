## Palindrome

```python
number = int(input("Enter a number: "))

original_number = number
reversed_number = 0

while number > 0:
    digit = number % 10
    reversed_number = reversed_number * 10 + digit
    number //= 10

if original_number == reversed_number:
    print(f"{original_number} is a Palindrome number.")
else:
    print(f"{original_number} is not a Palindrome number.")
```

## Prime Priniting

```python
print("Prime numbers from 1 to 100 are:")

for num in range(2, 101):  # Start from 2, as 1 is not a prime number
    is_prime = True
    for i in range(2, int(num**0.5) + 1):  # Check divisors up to the square root of num
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        print(num, end=" ")
```

```python
print("Prime numbers from 1 to 100 are:")

for n in range(2, 101):  
    if n>1:
        for i in range(2, n//2+1):
            if(n%i) == 0:
                break
            else:
                print(n)
```
