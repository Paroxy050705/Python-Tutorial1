```python
class ReportCard:
    def input(self):
        self.name = input("Enter your name: ")
        self.roll = input("Enter your Roll no: ")
        self.ben = int(input("Enter your Bengali marks: "))
        self.eng = int(input("Enter your English marks: "))
        self.math = int(input("Enter your Math marks: "))
        self.che = int(input("Enter your Chemistry marks: "))
        self.phy = int(input("Enter your Physics marks: "))
        self.bio = int(input("Enter your Biology marks: "))

    def calculate(self):
        self.total = self.ben + self.eng + self.math + self.che + self.phy + self.bio
        self.average = self.total / 6
        
        if self.average >= 75:
            self.grade = "A++"
        elif self.average >= 60:
            self.grade = "A"
        elif self.average >= 50:
            self.grade = "B"
        elif self.average >= 40:
            self.grade = "C"
        else:
            self.grade = "F (Fail Hahahaa!!!)"

    def displayOutput(self):
        print("\n=== Report Card ===")
        print("Name: ", self.name)
        print("Roll: ", self.roll)
        print("Bengali Marks: ", self.ben)
        print("English Marks: ", self.eng)
        print("Math Marks: ", self.math)
        print("Chemistry Marks: ", self.che)
        print("Physics Marks: ", self.phy)
        print("Biology Marks: ", self.bio)
        print("Total Marks: ", self.total)
        print("Average Marks: {:.2f}".format(self.average))
        print("Grade: ", self.grade)

student1 = ReportCard()
student1.input()
student1.calculate()
student1.displayOutput()
```

```python
# write a programm to create employee database, with emloyee id, name and basic salary as inputs, calculate DA(20% of Basic salary), 
# HRA(40% of basic salary) and gross salary(total), if the gross salary is > 60,000, than tax is 30% of gross salary
# otherwise 10% of gross salary

class Employee:
    def inputs(self):
        self.name = input("Enter your name: ")
        self.id = input("Enter your id: ")
        self.basicSalary = float(input("Enter your Basic Salary: "))

    def calculation(self):
        self.DA = self.basicSalary*(20/100)
        self.HRA = self.basicSalary*(40/100)
        self.grossSalary = self.basicSalary + self.DA + self.HRA

        if self.grossSalary > 60000:
            self.tax = self.grossSalary*(30/100)
        else:
            self.tax = self.grossSalary*(10/100)

    def outputs(self):
        print("Name: ", self.name)
        print("ID: ", self.id)
        print("DA: ", self.DA)
        print("HRA: ", self.HRA)
        print("Gross Salary: ", self.grossSalary)
        print("Tax: ", self.tax)
        print("Salary after tax: ", self.grossSalary - self.tax)

obj1 = Employee()
obj1.inputs()
obj1.calculation()
obj1.outputs()
```

```python
# write a programm to find factorial

class Factorial:
    def input(self, x):
        self.n = x

    def findFact(self):
        self.f = 1
        while(self.n > 1):
            self.f = self.f * self.n
            self.n = self.n - 1

    def output(self):
        return(self.f)


obj = Factorial()
p = int(input("Enter a no. "))
obj.input(p)
obj.findFact()
q = obj.output()
print("Factorial value: ", q)
```

```python
# add two numbers

class Addition:
    def input(self):
        self.x = int(input("Enter a no. "))
        self.y = int(input("Enter 2nd no. "))

    def calculate(self):
        self.z = self.x + self.y

    def output(self):
        return(self.z)


obj = Addition()
obj.input()
obj.calculate()
z = obj.output()
print("Sum = ", z)
```
