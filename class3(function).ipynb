```python
# write a program to create student database using key argument function with attribute - student id, student name and student marks in 3 different subjects and find the total and the avarage
def student_info(student_id, name, marks_phy, marks_che, marks_math):
    print("Student id = ", student_id)
    print("Student Name = ", name)
    print("Physics = ", marks_phy, "Chemistry = ", marks_che, "Math = ", marks_math)
    total = marks_phy + marks_che + marks_math
    avg = total/3
    print("Total = ", total)
    print("Avarage = ", avg)


print("Enter student details:")
ID = int(input("Enter Student ID: "))
N = input("Enter Student Name: ")
phy = int(input("Enter Marks in Physics: "))
che = int(input("Enter Marks in Chemistry: "))
math = int(input("Enter Marks in Mathematics: "))

student_info(student_id=ID, name=N, marks_phy=phy, marks_che=che, marks_math=math)
```

```python
# function of functions isn/t supported in python
# we use lambda for function of function in python
# lamda isn't a real function, it just gives a illusion of function

def add(x,y, m, n):
    z = x+y
    r = lambda m,n : m+n
    p = r(m,n)
    print("z = ", z)
    print("p = ", p)

add(3,2,9,1)
```

```python
# write a programm to add two numbers using object

class A:
    def input(self):
        self.x = int(input("Enter a number: "))
        self.y = int(input("Enter 2nd number: "))

        self.z = self.x+ self.y
    def output(self):
        return(self.z)

obj = A()
obj.input()
z = obj.output()
print("sum = ", z)
```
